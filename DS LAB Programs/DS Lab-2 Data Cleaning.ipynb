```python
#1. Handling missing values
#Mean/Mode/Median imputation

import pandas as pd
import numpy as np

df=pd.DataFrame({'Age':[25,30,np.nan,40,35],
    'Department':["HR","Finance","Finance",np.nan,'IT']})

#Display Original DataSet
print("Original DataSet(With Missing Values): ")
print(df,"\t")

#Using Mean
df['Age']=df['Age'].fillna(df['Age'].mean())

#Using Mode
df['Department']=df['Department'].fillna(df['Department'].mode()[0]);

#To Display Dataset After Handling Missing Values:
print("\nDataset After Handling Missing Values: ")
print(df,"\t")
```

```python
#2. Forward/Backward Fill

import pandas as pd
import numpy as np

df=pd.DataFrame({'Age':[25,30,np.nan,40,35],
    'Department':["HR","Finance","Finance",np.nan,'IT']})

#Display Original DataSet
print("Original DataSet(With Missing Values): ")
print(df,"\t")

#Forward Fill (use previous value)
print("\nDataset After using Forward Fill ")
df_ffill = df.copy()
df_ffill.ffill(inplace=True)
print(df_ffill)

#Forward Fill (use next value)
print("\nDataset After using Backward Fill ")
df_bfill = df.copy()
df_bfill.bfill(inplace=True)
print(df_bfill)
```

```python
#3. Deletion(drop rows or columns)

import pandas as pd
import numpy as np

df=pd.DataFrame({'Age':[25,30,np.nan,40,35],
    'Department':["HR","Finance","IT","Finance","IT"]})

#Display Original DataSet
print("Original DataSet(With Missing Values): ")
print(df,"\t")


# Delete rows with missing values

df_drop_rows=df.dropna()
print("\nAfter Dropping The Rows: \n",df_drop_rows)

# Delete Cloumns with missing values

df_drop_cols=df.dropna(axis=1)
print("\nAfter Dropping The Cols: \n",df_drop_cols)
```

```python
#4(a). Remove Duplicate Rows

import pandas as pd

# Create DataFrame
df = pd.DataFrame({
    'ID': [1, 2, 3, 4, 5, 6],
    'Name': ['Alice', 'Bob', 'Charlie', 'Alice', 'David', 'Bob'],
    'Age': [23, 25, 28, 23, 30, 25]
})

# Display Original Dataset
print("Original Data:")
print(df)

# Remove duplicate rows based on Name and Age
df_clean = df.drop_duplicates(subset=['Name', 'Age'])

# Display Cleaned Dataset
print("\nData After Removing Duplicates:")
print(df_clean)
```

```python
#4b. Create DataFrame with different date formats

import pandas as pd

df = pd.DataFrame({
    "Date": ['2025-01-05', '05/01/2025', 'Jan 5, 2025', '2025.01.05']
})

# Display original data
print("Original Data:")
print(df)

# Convert all dates to ISO 8601 format (YYYY-MM-DD)
df["Date"] = pd.to_datetime(df["Date"], dayfirst=True, errors="coerce").dt.strftime("%Y-%m-%d")

# Display standardized data
print("\nAfter Date Standardization (ISO 8601):")
print(df)
```

```python
# 4c. Correct Inconsistent Formats and Standardize Date
import pandas as pd
df=pd.DataFrame({
    "Date":['2025-01-05','05/01/2025','jan 5,2025','2025.01.05']
})
print("Original Data:\n",df)
#print("\n")

df['Date']=pd.to_datetime(df['Date'],errors='coerce').dt.strftime('%y-%m-%d')
print("\n",df)
```

```python
#4d. Case Normalization (converting text to lowercase and uppercase)
import pandas as pd

# Create DataFrame
df = pd.DataFrame({
    "Department": ["CSE", "cse", "CsE", "ECE", "ece"]
})

# Display original data
print("Original Data:")
print(df)

# Convert to lowercase
df["Lower Case"] = df["Department"].str.lower()

# Convert to uppercase
df["Upper Case"] = df["Department"].str.upper()

# Display normalized data
print("\nAfter Case Normalization:")
print(df)
```
