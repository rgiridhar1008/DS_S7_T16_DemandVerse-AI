```python
import matplotlib.pyplot as plt
y1=[] #stores +ve values
y2=[] #stores -ve values
x=range(-100,100,10) #Generates values from -100 to 90 with an interval of 10(-100,-90,-80..)
for i in x:y1.append(i**2) #y=x^2
for i in x:y2.append(-i**2) 
plt.plot(x,y1)
plt.plot(x,y2)
plt.xlabel("x")
plt.ylabel("y")
plt.ylim(-2000,2000) #yaxis limit
plt.axhline(0) #horizontal line
plt.axvline(0) #vertical line
plt.savefig("quad.png")
plt.show()
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
iris = sns.load_dataset("iris")
setosa = iris.loc[iris.species == "setosa"]

sns.kdeplot(
    x=setosa.petal_length,
    y = setosa.petal_width,
    fill = True
)
plt.title("KDE Plot of Setosa petal dimensions")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")

plt.show()
```

```python
import dash
from dash import dcc, html, Input, Output

app = dash.Dash(__name__)

app.layout = html.Div([
    html.Div("Square Calculator"),

    dcc.Input(
        id="input",
        type="text",
        placeholder="Enter a Number"
    ),

    html.Div(
        id="output"
    )
])


@app.callback(
    Output("output", "children"),
    Input("input", "value")
)
def update_value(input_data):
    try:
        return str(float(input_data) ** 2)
    except:
        return "Error: The input is not a number"


app.run(debug=False, port=8051)
```
